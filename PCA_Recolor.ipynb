# PCA Recolor
___

```python
#Library Import
import numpy as np
import random
from sklearn.decomposition import PCA
from PIL import Image
import matplotlib.pyplot as plt
```

#### Code

```python
# Load the image
sample_image = Image.open('Parthenon.jpg')
sample_image = np.array(sample_image, dtype=np.float64) / 255.0  # Normalize the image to [0, 1]

# Get the shape of the image
w, h, d = original_shape = tuple(sample_image.shape)

# Flatten the image into a 2D array (rows: pixels, columns: color channels)
image_array = np.reshape(sample_image, (w * h, d))

# Perform PCA for 3-1 reduction (reduce to 1 component)
pca_3_1 = PCA(n_components=1)
colour_components_3_1 = pca_3_1.fit_transform(image_array)
reconstructed_image_3_1 = pca_3_1.inverse_transform(colour_components_3_1)

# Perform PCA for 3-2 reduction (reduce to 2 components)
pca_3_2 = PCA(n_components=2)
colour_components_3_2 = pca_3_2.fit_transform(image_array)
reconstructed_image_3_2 = pca_3_2.inverse_transform(colour_components_3_2)

# Reshape the reconstructed arrays back into the original image shape
reconstructed_image_3_1 = np.reshape(reconstructed_image_3_1, original_shape)
reconstructed_image_3_2 = np.reshape(reconstructed_image_3_2, original_shape)

# Show the original and the PCA-reconstructed images after 3-1 and 3-2 reductions
plt.figure(figsize=(18, 6))

# Display the original image
plt.subplot(1, 4, 1)
plt.imshow(sample_image)
plt.title("Original Image")
plt.axis("off")

# Display the reconstructed image after 3-1 reduction
plt.subplot(1, 4, 2)
plt.imshow(reconstructed_image_3_1)
plt.title("3-1 PCA Reduced Image")
plt.axis("off")

# Display the reconstructed image after 3-2 reduction
plt.subplot(1, 4, 3)
plt.imshow(reconstructed_image_3_2)
plt.title("3-2 PCA Reduced Image")
plt.axis("off")

# Show the differences between 3-1 and 3-2 reductions
difference_3_1_3_2 = np.abs(reconstructed_image_3_1 - reconstructed_image_3_2)

plt.subplot(1, 4, 4)
plt.imshow(difference_3_1_3_2)
plt.title("Difference (3-1 vs 3-2)")
plt.axis("off")

plt.show()
```
